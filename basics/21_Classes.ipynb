```python
class LukeList:
    def __init__(self):
        self._items = []

    def add(self, item):
        self._items.append(item)

    def __getitem__(self, index):
        return self._items[index]

    def __setitem__(self, index, value):
        self._items[index] = value

    def __repr__(self):
        return str(self._items)

    def __len__(self):
        return len(self._items)
```

```python
my_list = LukeList()
```

```python
my_list.add("data nerd")
```

```python
my_list.add("data analyst")
```

```python
class BaseSalary:
    def __init__(self, base_salary, bonus_rate = 0.1, symbol = '$'):
        self.base_salary = base_salary
        self.bonus_rate = bonus_rate
        self.symbol = symbol
        self.total_salary = base_salary * (1 + bonus_rate)
        self.bonus = self.total_salary - self.base_salary

    def __repr__(self):
        return f"{self.symbol}{self.base_salary:,.0f}"

    def cal_sal(self):
        return f"{self.symbol}{self.total_salary:,.0f}"

    def cal_bonus(self):
        return f"{self.symbol}{self.bonus:,.0f}"
```

```python
salary = BaseSalary(100000)
```

```python
salary.symbol
```

```python
salary.__repr__()
```

```python
salary
```

```python
salary.cal_sal()
```

```python
salary.cal_bonus()
```
